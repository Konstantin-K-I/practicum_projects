### 1.2. Предобработка данных

<font color='#777778'>Проверьте названия столбцов в датасетах: все ли они точно отражают содержимое данных и оформлены в удобном для работы стиле. При необходимости приведите их к единому аккуратному стилю.</font>

В названии столбцов ` market ` и ` funding_total_usd ` находятся пробелы в начали и в конце, а в названии столбцов `round_` указаны заглавные буквы. При работе с таким названием столбцов могут возникнуть сложности, далее приведем к единому строчному виду без пробелов.
Названия столбцов в обоих датафреймах отражают содержимое и оформлены в аккуратном "змеином" стиле. Дополнительно переименовывать столбцы не требуется.

```python
#Переименовываю столбцы на названия без пробелов
cb_investments_df = cb_investments_df.rename(columns={' market ': 'market',
                        ' funding_total_usd ': 'funding_total_usd'})
```

```python
cb_investments_df.columns = cb_investments_df.columns.str.lower() #Меняю названия на нижний регистр

print(cb_investments_df.columns) #Проверяю названия столбцов
```

Переименование прошло успешно!

<font color='#777778'>Уберите в столбце `funding_total_usd` выделение разрядов и приведите его к числовому типу.</font>

```python
#Удаляю пробелы и запятые, оставляю только цифры
cb_investments_df['funding_total_usd'] = (cb_investments_df['funding_total_usd'].astype(str).str.strip().str.replace(',', '', regex=False).str.replace(' ', '', regex=False))

#Преобразую в число, а некорректные значения в NaN
cb_investments_df['funding_total_usd'] = pd.to_numeric(cb_investments_df['funding_total_usd'], errors='coerce')

#Преобразую данные к численному типу, но с сохранением пропусков за счет использования nullable-типа (пропуски останутся, но тип столбца сменится на числовой)
cb_investments_df['funding_total_usd'] = cb_investments_df['funding_total_usd'].astype('Int64')

print(cb_investments_df['funding_total_usd'].head(20)) #Проверяю значения в столбце
```

<div class="alert alert-block alert-success">✔️

Верное решение  использовать опцию errors='coerce'. Это позволит заменить неверные значения на NaT (Not a Time), что упростит дальнейшую работу с данными.

Смена столбца на числовой тип прозведен успешно!

<font color='#777778'>Определите компании с аномальным объёмом общего финансирования — используйте метод IQR отдельно по каждому сегменту. Напомним, что все нишевые сегменты должны быть объединены в одну группу, а средние — в другую.

Определите сегменты рынка с наибольшей долей компаний, получивших аномальное финансирование, и выведите топ таких сегментов.

```python
# Создаю пустой список для хранения аномальных компаний
anomalous_companies = []

# Создаю список для хранения статистики по сегментам
enumeration = []

#Ищу аномальные значения и компании 
for market_segment in cb_invest_clean['market'].unique():
    data = cb_invest_clean[cb_invest_clean['market'] == market_segment]
    funding_data = data['funding_total_usd']
    if len(funding_data) > 0:
        Q1 = funding_data.quantile(0.25)
        Q3 = funding_data.quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        #Нахожу аномальные значения
        anomalies = data[(data['funding_total_usd'] < lower_bound) | 
                        (data['funding_total_usd'] > upper_bound)]
        
        #Сохраняю аномальные компании
        for index, row in anomalies.iterrows():
            anomalous_companies.append({
                'company': row['name'],
                'segment': market_segment,
                'funding': row['funding_total_usd']
            })
        
        #Собираю статистику по сегменту
        anomalies_counts = len(anomalies)
        total_counts = len(funding_data)
        
        if total_counts > 0:
            relative = round((anomalies_counts / total_counts) * 100, 2)
        else:
            relative = 0
            
        enumeration.append({
            'segment': market_segment,
            'total': total_counts,
            'anomalies': anomalies_counts,
            'relative': relative
        })

#Создаю дф с аномальными компаниями
anomalous_companies_df = pd.DataFrame(anomalous_companies)

#Вывожу список аномальных компаний
print('Компании с аномальным финансированием:')
print(anomalous_companies_df)

#Создаю дф со статистикой по сегментам
anomalies_list_df = pd.DataFrame(enumeration)

#Сортирую сегменты по относительной доле аномалий
top_segments = anomalies_list_df.sort_values('relative', ascending=False)

#Проверяю данные
print('\nСегменты с аномальным объёмом финансирования:')
print(top_segments.head())
```
